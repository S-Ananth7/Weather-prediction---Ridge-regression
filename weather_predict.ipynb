```python
from sklearn.linear_model import Ridge
reg = Ridge(alpha=.1)
```

```python
predictors = ["precip","temp_max", "temp_min","snow","snow_depth"]
```

```python
train = core_weather.loc[:"2020-12-31"]
```

```python
test = core_weather.loc["2021-01-01":]
```

```python
reg.fit(train[predictors], train["target"])
```

```python
predictions = reg.predict(test[predictors])
predictions
```

```python
from sklearn.metrics import mean_absolute_error
```

```python
mean_absolute_error(test["target"], predictions)
```

```python
combined = pd.concat([test["target"], pd.Series(predictions, index=test.index)], axis=1)
combined.columns=["actual", "predictions"]
```

```python
combined
```

```python
combined.plot()
```

```python
reg.coef_
```

```python
def create_predictions(predictors, core_weather, reg):
    train = core_weather.loc[:"2020-12-31"]
    test = core_weather.loc["2021-01-01":]
    reg.fit(train[predictors], train["target"])
    predictions = reg.predict(test[predictors])
    error = mean_absolute_error(test["target"], predictions)
    combined = pd.concat([test["target"], pd.Series(predictions, index=test.index)], axis=1)
    combined.columns = ["actual", "predictions"]
    return error, combined
```

```python
core_weather["month_max"] = core_weather["temp_max"].rolling(30).mean()
```

```python
core_weather
```

```python
core_weather["month_day_max"] = core_weather["month_max"]/core_weather["temp_max"]
core_weather
```

```python
core_weather["max_min"] = core_weather["temp_max"] / core_weather["temp_min"]
core_weather
```

```python
predictors = ["precip", "temp_max", "temp_min", "month_max", "month_day_max", "max_min"]
```

```python
core_weather = core_weather.iloc[30:,:].copy()
core_weather
```

```python
import numpy as np

core_weather = core_weather.replace([np.inf, -np.inf], np.nan)
core_weather = core_weather.ffill()
core_weather = core_weather.dropna(subset= predictors + ["target"])
```

```python
error, combined = create_predictions(predictors, core_weather, reg)
error
```

```python
combined.plot()
```

```python
core_weather["monthly_avg"] = core_weather["temp_max"].groupby(core_weather.index.month).transform(lambda x: x.expanding(1).mean())
```

```python
core_weather["day_of_year_avg"] = core_weather["temp_max"].groupby(core_weather.index.day_of_year).transform(lambda x: x.expanding(1).mean())
```

```python
predictors = ["precip","temp_max","temp_min", "month_day_max","max_min","day_of_year_avg", "monthly_avg"]
```
